Run this notebook in an environment containing `pele-python`. The renderer needs no Node.js, browser, IPython import, or setup cell. The notebook frontend itself supplies rich SVG display.

## Large tree (63 nodes)

Generate a balanced binary tree with **63 nodes and 62 edges**. Each node after the root connects to its parent (`node // 2`). The left-to-right layout and fixed SVG size keep the labels readable; scroll through the output to explore all six levels.


```python
from pele_python import render

node_count = 63
lines = ["flowchart LR", f"  accTitle: Binary tree with {node_count} nodes"]
lines.extend(f'  N{node}["Node {node}"]' for node in range(1, node_count + 1))
lines.extend(f"  N{node // 2} --> N{node}" for node in range(2, node_count + 1))
tree_source = "\n".join(lines)

tree = render(
    tree_source,
    idPrefix='notebook-tree',
    fontSize=14,
    autoDirection=False,
    responsive=False,
)
tree
```

## Save or reuse the markup

`diagram.svg` contains the already rendered SVG. For plain strings, use `render_svg(source)`. To save: `Path('diagram.svg').write_text(diagram.svg, encoding='utf-8')`.
